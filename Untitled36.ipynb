```python
import pandas as pd
df=pd.read_csv("/content/creditcard.csv");
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df.drop("Time",axis=1,inplace=True);
```

```python
for i in df['Class'].unique():
  print(i,"totalcount=",(df["Class"]==i).sum());
```

```python
print(df.info())
```

```python
from imblearn.over_sampling import SMOTE
X = df.drop("Class", axis=1)
y = df["Class"]
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x="Class", data=df)
plt.title("Class Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt;
for i in df.columns:
  plt.boxplot(df[i])
  plt.show()
```

```python
cnt=0
cn=0
for i in y_resampled:
    if i==0:
      cnt=cnt+1;
    else:
      cn=cn+1
print(f" totla 0 count is={cnt}  and totla 1  count is{cn}");
```

```python
from sklearn.model_selection import train_test_split;
from sklearn.ensemble import RandomForestClassifier;
```

```python
xtr,xts,ytr,yts=train_test_split(X_resampled,y_resampled,test_size=0.2,random_state=42);
```

```python
model=RandomForestClassifier(max_depth=3,random_state=42);
```

```python
model.fit(xtr,ytr);
```

```python
pred=model.predict(xts);
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score;
```

```python
print("confusion_metrix")
print(confusion_matrix(pred,yts))
```

```python
print("accuracy_score")
print(accuracy_score(pred,yts))
print("precession");
print(precision_score(pred,yts))
print("recall")
print(recall_score(pred,yts))
```
